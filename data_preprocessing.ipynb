Text cleaning and standardization

```python
bookings = pd.read_csv("Bookings.csv")
```

```python
bookings['Customer_Name'] = (
    bookings['Customer_Name'].str.lower()
)
```

```python
bookings['Customer_Name'] = (
    bookings['Customer_Name'].str.title()
)
```

Creating calculated columns

Dataset merging using common keys

```python
bookings = pd.read_csv("Bookings.csv")
movies = pd.read_csv("Movies.csv")
shows = pd.read_csv("Shows.csv")
```

```python
booking_shows = pd.merge(
    bookings,
    shows,
    on='Show_ID',
    how='inner'
)

display(booking_shows.head())
```

```python
booking_shows_left = pd.merge(
    bookings,
    shows,
    on='Show_ID',
    how='left'
)
```

```python
final_data = pd.merge(
    bookings,
    shows,
    on='Show_ID',
    how='inner'
)

final_data = pd.merge(
    final_data,
    movies,
    on='Movie_ID',
    how='inner'
)

display(final_data.head())
```

```python
print(final_data.shape)

print(final_data.columns)

final_data.info()
```

Splitting and recombining datasets

```python
part1 = bookings.iloc[:18000]

part2 = bookings.iloc[18000:]

print(part1.shape)

print(part2.shape)
```

```python
combined_data = pd.concat(
    [part1, part2],
    axis=0
)

print(combined_data.shape)
```
